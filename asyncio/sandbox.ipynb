```python
import asyncio
from time import sleep, perf_counter
```

```python
def fun(x: int=1):
    "Synchrone Funktion mit Wartedauer x"
    sleep(x)
    return x
async def asfun(x: int=1):
    "Asynchrone Funktion mit Wartedauer x"
    await asyncio.sleep(x)
    return x
```

```python
async def foo(x,y):
    # async funktionen zusammen initiieren
    a,b = await asyncio.gather(asfun(1),asfun(2))
    return a+b
#
t0 = perf_counter()
res = await foo(1,2) # 2s
print(res,perf_counter()-t0)
```

```python
async def foo(x,y):
    # async funktionen zusammen initiieren
    a,b = await asyncio.gather(asfun(x),asfun(y))
    c,d = await asyncio.gather(asfun(x),asfun(y))
    return a+b+c+d
#
t0 = perf_counter()
res = await foo(1,2) # 2s+2s
print(res,perf_counter()-t0)
```

```python
async def foo(x,y):
    a,b = await asyncio.gather(asfun(1),fun(2))
    return a+b
#
await foo(1,2) # TypeError: An asyncio.Future, a coroutine or an awaitable is required
```

```python
def foo(x,y):
    a = fun(x)
    b = await asfun(y)
    return a+b
#
foo(1,2) # SyntaxError: 'await' outside async function
```

```python
async def foo(x,y,z):
    a = fun(x) # 1s
    b,c = await asyncio.gather(asfun(y),asfun(z)) # 3s
    return a+b+c
#
await foo(1,2,3) # 4s
```
