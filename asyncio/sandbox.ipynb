```python
import asyncio
from time import sleep
```

```python
def fun(x: int=1):
    sleep(x)
    return x
```

```python
async def asfun(x: int=1):
    await asyncio.sleep(x)
    return x
```

```python
async def foo(x,y):
    a,b = await asyncio.gather(asfun(1),asfun(2))
    return a+b
#
await foo(1,2) # 2s
```

```python
def foo(x,y):
    a = fun(x)
    b = asyncio.run(asfun(y))
    return a+b
#
foo(1,2) # 2s
```

```python
async def foo(x,y):
    a,b = await asyncio.gather(asfun(1),fun(2))
    return a+b
#
await foo(1,2) # TypeError: An asyncio.Future, a coroutine or an awaitable is required
```

```python
async def foo(x,y):
    a = await asfun(x)
    b = await asfun(y)
    return a+b
#
await foo(1,2) # 3s
```

```python
async def foo(x,y):
    a = fun(x)
    b = await asfun(y)
    return a+b
#
await foo(1,2) # 3s
```

```python
def foo(x,y):
    a = fun(x)
    b = await asfun(y)
    return a+b
#
foo(1,2) # SyntaxError: 'await' outside async function
```

```python
async def foo(x,y,z):
    a = fun(x) # 1s
    b,c = await asyncio.gather(asfun(y),asfun(z)) # 3s
    return a+b+c
#
await foo(1,2,3) # 4s
```
