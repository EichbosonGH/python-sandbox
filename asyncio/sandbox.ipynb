```python
import asyncio
from time import time
```

```python
async def asfun(x: int=1):
    await asyncio.sleep(x)
    return x
```

```python
t0 = time()
a,b = await asyncio.gather(
    asfun(1),
    asfun(2)
    )
print(time()-t0,a+b)
```

```python
# 
a = await asfun(1)
b = await asfun(2)
print(a+b)
```

```python
a,b = await asfun(1),await asfun(2)
print(a+b)
```

```python
async def foo(x,y):
    a = await asfun(x=x)
    b = await asfun(y)
    return a+b
#
await foo(1,2)
```
